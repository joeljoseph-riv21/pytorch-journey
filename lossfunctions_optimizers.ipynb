<h3>Loss Functions and Optimizers</h3>

```python
import torch
import torch.nn as nn
import torch.optim as optims

#model creation in a sequential form
model = nn.Sequential(
    nn.Linear(8,3),
    nn.ReLU(),
    nn.Linear(3,4),
    nn.ReLU(),
    nn.Linear(4,1)
)
```

```python
#loss functions and optimizer creation
criterion=nn.MSELoss()
optimizer=optims.AdamW(model.parameters(), lr=0.001, weight_decay=0.01)
```

```python
#creating datas for comaprison of criterion
ans=torch.randn(4,8)
key=torch.randn(4,1)
```

```python
#training begins
optimizer.zero_grad()
answers=model(ans)
loss=criterion(answers, key)
loss.backward()
optimizer.step()

#checking shapes
print(ans.shape)
print(key.shape)
print(f"Target: {answers.shape} Loss: {loss.item():.4f}")
```
